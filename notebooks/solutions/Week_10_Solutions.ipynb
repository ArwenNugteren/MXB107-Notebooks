### **Demonstrating Example**

A phone company advertises that the average battery life of their phones (when continuously watching videos), denoted as $\mu$, is 8 hours.

To verify this claim, an independent random sample of 31 phones was tested.

````r
smpl_data = c(7.2, 8.53, 8.07, 7.99, 7.79, 7.77, 8.9, 7.64, 7.35, 8.45, 9.14, 7.93, 7.35, 7.52, 7.41, 8.27,
7.55, 7.5, 8.53, 8.37, 8.17, 8.15, 8.02, 7.63,7.64, 8.83, 8.17, 7.41, 7.7, 8.21, 7.13)
````

**Write down the null and alternative hypotheses for testing whether the company’s claim is correct.**


Since we are testing whether or not there is evidence *against* the company’s claim that the average battery life is 8 hours, the alternative hypothesis should challenge this claim.  

Because we are not concerned if the battery lasts longer than 8 hours (that would be favorable to consumers), we only test if it is **less** than 8 hours.  

$$
\begin{align}
H_0: \mu &= 8 \\
H_1: \mu &< 8
\end{align}
$$

This is a **left-tailed test** (or often simply referred to as one-tailed or one-sided test) of the mean.

**Approximate the sampling distribution of the sample mean under the null hypothesis.**

Since our sample size is larger than 30, it is reasonable to assume the sample mean approximately has a sampling distribution that is normal.  We do not know the population variance of battery life, but we can estimate it from the sample.

````r
var(smpl_data)
sd(smpl_data)

length(smpl_data)

mean(smpl_data)
````

$$\bar{x} \sim \mathcal{N}\Big(7.95,\frac{0.267}{31}\Big)$$

**Define the z-test statistic for testing the null hypothesis**


$$
Z = \frac{\bar{x} - \mu_0}{\frac{s}{\sqrt{n}}} \approx \frac{7.95 - 8}{\frac{0.517}{\sqrt{31}}} \approx -0.538
$$


**Now we want to compute the p-value**

Since we have a left-tailed test, the p-value is given by
$$
P(Z < -0.538 | H_0 \mbox{ is true})
$$

This is the probability of getting our observed test statistic value or more extreme (in the direction of the alternative) given that the null is true.  If this probability is small, it would provide evidence against the null in favour of the alternative.

Since $Z$ is approximately standard normal, we can compute the p-value using the cdf of the standard normal distribution (we need R to help us do that).  The full set of R code to conduct the hypothesis test is below.  

````r
xbar = mean(smpl_data)
s = sd(smpl_data)
n = 31
# compute the test statistic
Z = (xbar-8)/(s/sqrt(n))
Z

p = pnorm(Z)
p
````

Here we see the p-value is quite large, around 0.28. Thus it was reasonably likely to get the sample we got (or more extreme) if the null was true.  Thus there is insufficient evidence against the null.  That is, from this data, there is no evidence against the assumption that the mean battery life is 8 hours.

#### **Question 1**

Is there evidence of a difference in the average city mileage between cars manufactured in 2015 and 2020?  Interpret the results for a non-statistician stakeholder.

````r
epa_data = read.csv("./datasets/epa_data.csv")
epa_data %>% str()
````

<details>
<summary>▶️ Click to show the solution</summary>

```r
# define mu_2015 and mu_2020 as the true mean city mileage for cars manufactured in 2015 and 2020, respectively
# the null and alternative hypotheses are

#H0: mu_2020 = mu_2015
#H1: mu_2020 != mu_2015

# pull out relevant data
mileage_2015 = epa_data$city[epa_data$year == 2015]
mileage_2020 = epa_data$city[epa_data$year == 2020]

# check if sample sizes are large enough for normal approximation
length(mileage_2015) > 30
length(mileage_2020) > 30

# both sample sizes greater 30, so all good for the normal approximation


# compute the test statistic
z = (mean(mileage_2020) - mean(mileage_2015))/sqrt(var(mileage_2020)/length(mileage_2020) + var(mileage_2015)/length(mileage_2015))
z

# since this is a two-sided test, we need to compute P(|Z| > |z|), which we can do as follows (since the normal distribution is symmetric)
p = 2*pnorm(z, lower.tail = FALSE)
p

# at around 0.00015, the p-value is very small.  Hence there is overwhelming evidence against the null, in favour of the alternative
# Thus there is overwhelming evidence that the average city mileage differs between 2015 and 2020
```

#### **Question 1.3**

Is there evidence that the proportion of Star Trek: The Next Generation episodes that pass the Bechdel-Wallace Test is different from 0.4? Interpret the results for a non-statistician stakeholder.


**Note**: While these series have ended and you technically have the full “population” of some values (e.g., results of the Bechdel-Wallace test), we still ask you to test whether or not the proportion of episodes that pass the test is equal to 0.4. This may seem counter-intuitive, but you can think of it as follows:

- The episode test results are treated as realisations from an unknown probability distribution $f$ (here, Bernoulli(p)).
- Although the episodes are released, we are interested in the underlying process that generates these values. This includes not-yet-released episodes or hypothetical similar episodes. Simply examining the “complete” population of Bechdel test results is not sufficient; instead, we rely on a statistical model to quantify uncertainty.

<details>
<summary>▶️ Click to show the solution</summary>

```r
# Define p as the true proportion of TNG episodes that pass the test
# The null and alternative hypotheses are as follows

#H0: p = 0.4
#H1: p != 0.4

# pull out observed proportion of episodes passing test
TNG_tests = episodes$Bechdel.Wallace.Test[episodes$Series == "TNG"]
n = length(TNG_tests)
n
p_obs = mean(TNG_tests)
p_obs

# check if normal approximation is ok
n*0.4
n*0.6

# both np > 10 and n(1-p) > 10 so we are ok for the normal approximation

# compute the test statistic
z = (p_obs - 0.4)/sqrt(0.4*0.6/n)
z

# compute the p-value for two sided test
pval = 2*pnorm(z, lower.tail = FALSE)
pval

# at around 0.3 the pvalue is quite large, thus there is insufficient evidence against the null
# Thus there is no evidence that the proportion of TNG episodes that pass the test is different to 0.4
```
