### **Demonstrating Example**

A phone company advertises that the average battery life of their phones (when continuously watching videos), denoted as $\mu$, is 8 hours.

To verify this claim, an independent random sample of 31 phones was tested.

````r
smpl_data = c(7.2, 8.53, 8.07, 7.99, 7.79, 7.77, 8.9, 7.64, 7.35, 8.45, 9.14, 7.93, 7.35, 7.52, 7.41, 8.27,
7.55, 7.5, 8.53, 8.37, 8.17, 8.15, 8.02, 7.63,7.64, 8.83, 8.17, 7.41, 7.7, 8.21, 7.13)
````

**Write down the null and alternative hypotheses for testing whether the company’s claim is correct.**


Since we are testing whether or not there is evidence *against* the company’s claim that the average battery life is 8 hours, the alternative hypothesis should challenge this claim.  

Because we are not concerned if the battery lasts longer than 8 hours (that would be favorable to consumers), we only test if it is **less** than 8 hours.  

$$
\begin{align}
H_0: \mu &= 8 \\
H_1: \mu &< 8
\end{align}
$$

This is a **left-tailed test** (or often simply referred to as one-tailed or one-sided test) of the mean.

**Approximate the sampling distribution of the sample mean under the null hypothesis.**

Since our sample size is larger than 30, it is reasonable to assume the sample mean approximately has a sampling distribution that is normal.  We do not know the population variance of battery life, but we can estimate it from the sample.

````r
var(smpl_data)
sd(smpl_data)

length(smpl_data)

mean(smpl_data)
````

$$\bar{x} \sim \mathcal{N}\Big(7.95,\frac{0.267}{31}\Big)$$

**Define the z-test statistic for testing the null hypothesis**


$$
Z = \frac{\bar{x} - \mu_0}{\frac{s}{\sqrt{n}}} \approx \frac{7.95 - 8}{\frac{0.517}{\sqrt{31}}} \approx -0.538
$$


**Now we want to compute the p-value**

Since we have a left-tailed test, the p-value is given by
$$
P(Z < -0.538 | H_0 \mbox{ is true})
$$

This is the probability of getting our observed test statistic value or more extreme (in the direction of the alternative) given that the null is true.  If this probability is small, it would provide evidence against the null in favour of the alternative.

Since $Z$ is approximately standard normal, we can compute the p-value using the cdf of the standard normal distribution (we need R to help us do that).  The full set of R code to conduct the hypothesis test is below.  

````r
xbar = mean(smpl_data)
s = sd(smpl_data)
n = 31
# compute the test statistic
Z = (xbar-8)/(s/sqrt(n))
Z

p = pnorm(Z)
p
````

Here we see the p-value is quite large, around 0.28. Thus it was reasonably likely to get the sample we got (or more extreme) if the null was true.  Thus there is insufficient evidence against the null.  That is, from this data, there is no evidence against the assumption that the mean battery life is 8 hours.

#### **Question 1.1**

Is there evidence that the average IMDB rating of Star Trek: The Original Series episodes is lower than 7.7? Interpret the results for a non-statistician stakeholder.

````r
# Let's firstly write down the null and alternative hypotheses.
# Define mu as the true mean IMDB rating of Star Trek: The Original Series episodes

#H0: mu = 7.7
#H1: mu < 7.7

# Let's extract the data that we need
IDMB_ratings = episodes$IMDB.Ranking[episodes$Series == "TOS"]

# check the sample size is large enough for normal approximation of sampling distribution of the test statistic
length(IDMB_ratings)
# 80

# since our sample size > 30 we are ok for normal approximation

# compute the test statistic
z = (mean(IDMB_ratings) - 7.7)/(sd(IDMB_ratings)/sqrt(length(IDMB_ratings)))
z
# -1.82206462086615

# now we want to compute the p-value, which will be P(Z < -1.82206462086615) since this is a left-tailed test
# we can compute this using the cdf of the standard normal distribution

p = pnorm(z)
p
# 0.0342225905572484

# the p-value is quite small, thus there is some evidence against H0, in favour of the alternative.
# Thus there is some evidence that the mean IMDB rating is less than 7.7.
````

#### **Question 1.2**


Is there evidence that the mean IMDB rating of episodes of Star Trek: The Original Series differs from the mean rating of Star Trek: The Next Generation? Interpret the results for a non-statistician stakeholder.

````r
# now we are testing if two population means are different
# Define mu_TOS and mu_TNG as mean ratings for The Original Series and The Next Generation, respectively

# The null and alternative hypotheses are given by:
#H0: mu_TOS = mu_TNG
#H1: mu_TOS != mu_TNG

# pull out the data we need
ratings_TOS = episodes$IMDB.Ranking[episodes$Series == "TOS"]
ratings_TNG = episodes$IMDB.Ranking[episodes$Series == "TNG"]

# check sample sizes
length(ratings_TOS)
length(ratings_TNG)

# both sample sizes > 30, ok for normal approximation

# compute the test statistic
z = (mean(ratings_TOS) - mean(ratings_TNG))/sqrt(var(ratings_TOS)/length(ratings_TOS) + var(ratings_TNG)/length(ratings_TNG))
z

# since this is a two-sided test, we need to compute P(|Z| > |z|), which we can do as follows (since the normal distribution is symmetric)
p = 2*pnorm(z, lower.tail = FALSE)
p

# at around 0.43, the p-value is quite large, hence there is insufficient evidence against the null hypothesis
# thus there is no evidence that the mean rating differs between The Original Series and The Next Generation

````

#### **Question 1.3**

Is there evidence that the proportion of Star Trek: The Next Generation episodes that pass the Bechdel-Wallace Test is different from 0.4? Interpret the results for a non-statistician stakeholder.


**Note**: While these series have ended and you technically have the full “population” of some values (e.g., results of the Bechdel-Wallace test), we still ask you to test whether or not the proportion of episodes that pass the test is equal to 0.4. This may seem counter-intuitive, but you can think of it as follows:

- The episode test results are treated as realisations from an unknown probability distribution $f$ (here, Bernoulli(p)).
- Although the episodes are released, we are interested in the underlying process that generates these values. This includes not-yet-released episodes or hypothetical similar episodes. Simply examining the “complete” population of Bechdel test results is not sufficient; instead, we rely on a statistical model to quantify uncertainty.

````r
# Define p as the true proportion of TNG episodes that pass the test
# The null and alternative hypotheses are as follows

#H0: p = 0.4
#H1: p != 0.4

# pull out observed proportion of episodes passing test
TNG_tests = episodes$Bechdel.Wallace.Test[episodes$Series == "TNG"]
n = length(TNG_tests)
n
p_obs = mean(TNG_tests)
p_obs

# check if normal approximation is ok
n*0.4
n*0.6

# both np > 10 and n(1-p) > 10 so we are ok for the normal approximation

# compute the test statistic
z = (p_obs - 0.4)/sqrt(0.4*0.6/n)
z

# compute the p-value for two sided test
pval = 2*pnorm(z, lower.tail = FALSE)
pval

# at around 0.3 the pvalue is quite large, thus there is insufficient evidence against the null
# Thus there is no evidence that the proportion of TNG episodes that pass the test is different to 0.4
````

#### **Question 1.4**


Is there evidence that the proportion of episodes that pass the Bechdel-Wallace Test differs between Star Trek: The Next Generation and Star Trek: Voyager? Interpret the results for a non-statistician stakeholder.

````r
#H0: p_TOS - p_TNG = 0
#H1: p_TOS - p_TNG != 0

episodes %>%
  filter(Series %in%  c("TOS", "TNG")) %>%
  group_by(Series) %>%
  summarise(p_hat = mean(Bechdel.Wallace.Test),
            n = n()) -> summaryStats

TNG_stats = summaryStats %>% filter(Series == "TNG")
TOS_stats = summaryStats %>% filter(Series == "TOS")

z_obs = (TOS_stats$p_hat - TNG_stats$p_hat)/sqrt(TOS_stats$p_hat*(1-TOS_stats$p_hat)/TOS_stats$n + TNG_stats$p_hat*(1-TNG_stats$p_hat)/TNG_stats$n)
z_975 = qnorm(0.975)

abs(z_obs) > z_975

#As abc(z_obs) > z_975, at 95% significance level, there is evidence against the null hypothesis that
#the proportion of episodes passing the Bechdel test of TOS is equal to that of TNG. Reject the null hypothesis in favour of the
#alternative one that these two proportions are different.
````

````r
#H0: p = 0.5
#H1: p > 0.5


epa_data %>%
  filter(year == 2010) %>%
  summarise(p_hat = mean(ifelse(trans == "Manual", 1, 0)),
            n = n(),
            p_H0 = 0.5,
            z_95 = qnorm(0.95)) %>%
  mutate(z_obs = (p_hat-p_H0)/sqrt(p_H0*(1-p_H0)/n))

#As z_obs < z_95, at 95% significance level, there is no evidence against the null hypothesis that
#the proportion of cars produced with manual transmissions in 2010 is 0.5. Do not reject the null hypothesis.
````

#### **Question 2.3**

Is there evidence that the proportion of cars produced with manual transmissions for the years 2011 and 2012 has decreased? Interpret the results for a non-statistician stakeholder.

````r
#H0: mu_2012 - mu_2011 = 0
#H1: mu_2012 - mu_2011 < 0

epa_data %>%
  filter(year %in%  c(2012, 2011)) %>%
  group_by(year) %>%
  summarise(p_hat = mean(ifelse(trans == "Manual", 1, 0)),
            n = n()) -> summaryStats

epa_2012 = summaryStats %>% filter(year == 2012)
epa_2011 = summaryStats %>% filter(year == 2011)

z_obs = (epa_2012$p_hat - epa_2011$p_hat)/sqrt(epa_2012$p_hat*(1-epa_2012$p_hat)/epa_2012$n + epa_2011$p_hat*(1-epa_2011$p_hat)/epa_2011$n)
z_05 = qnorm(0.05)

z_obs < z_05

#As z_obs > z_05, at 95% significance level, there is no evidence against the null hypothesis that the proportion of cars
# produced with manual transmissions for the years 2011 and 2012 has remained the same. Do not reject the null hypothesis.
````
