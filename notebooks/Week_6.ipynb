## Example: Ice-cream shop orders
An icecream shop serves customers one at a time. They can order one flavour out of the options: chocolate, vanilla, coffee, pistachio, strawberry and gumdrop. They also have an option to add toppings.



### Scenario 1
The next customer is a toddler who demands to choose their own icecream flavour, but does not understand what the different flavours are. We can assume that each icecream flavour has an equal probability of being chosen.

What type of random variable is the toddler's choice of flavour? What is the probability that they choose pistachio?

<details> <summary>▶️ Click to show the solution</summary>

As each choice is equally likely, we can model this using the discrete uniform distribution. There are 6 options, so the probability of the toddler choosing pistachio is $\frac16$.

</details>

### Scenario 2
The toddler has an 8 year old sibling who has been promised that they can order sprinkles if they have been well-behaved. Unfortunately, they do not know if their parents are aware that they tidied up their room by stuffing everything under the bed. They estimate that they were very sneaky in doing so, and that the probability that their parents believe they have been good as 0.8.

Write down the probability distribution for getting sprinkles. What is the variance of this distribution?

<details> <summary>▶️ Click to show the solution</summary>

Let $S=1$ represent the child getting sprinkles, and $S=0$ represent them missing out. Since there are only two outcomes, $S$ can be modelled as Bernoulli random variable with parameter $p=0.8$.

For the Bernoulli distribution, we know $\Var(S) = p(1-p) = 0.8 * 0.2 = 0.16$.

</details>

### Scenario 3
While you are eating your icecream (coffee with chocolate sauce on top), you are suprised to notice multiple people ordering vanilla. You estimate that the probability of ordering vanilla is 0.36.

As you continue to eat your icecream, you observe 20 more customers ordering. Let $X$ be the random variable representing whether they order vanilla.

a) What is the distribution of $X$? Write down the pmf of $X$.

b) What is the probability that exactly 4 people order vanilla? What about more than 17 people?

c) What is the expected number of vanilla orders?

<details> <summary>▶️ Click to show the solution</summary>

a) Let $X$ be the number of customers who order vanilla. Then $X \sim \text{Bin}(20, 0.8)$. It has pmf:
$$Pr(X = k) = \binom{20}k (0.36)^k (0.64)^{20-k} $$

b) We need to compute $Pr(X=4)$.

$$P(X=4) = \binom{20}4 (0.36)^4 (0.64)^{16} \approx 0.645$$

\begin{align*} P(X > 17) & = P(X = 18) + P(X=19) + P(X=20) \\
& = \binom{20}{18} 0.36^{18} 0.64^2 + \binom{20}{19} 0.36^{19} 0.64 + \binom{20}{20} 0.36^{20} \\
& \approx 8.52 \times 10^{-7}
\end{align*}

c) For the binomial distribution, $\mathbb{E}(X) = np = 20 \cdot 0.36 = 7.2$, so the expected number of vanilla orders is 7.2.
</details>

### Scenario 4
Customers order toppings independently with probability 0.4. Let $Y$ be the number of customers up till and including the first customer of the day who orders a topping.

a) Compute the probability that $Y=4$ and $Y \geq 3$.

b) Calculate $\mathbb{E}(Y)$ and interpret this value.

<details> <summary>▶️ Click to show the solution</summary>

We can model $Y$ as a geometric random variable with parameter $p = 0.4$, giving us the pmf $P(Y = k) = 0.6^{k-1} 0.4$.

a) $$Pr(Y = 4)  = 0.6^3 \cdot 0.4 = 0.0864$$

\begin{align*} Pr(Y \geq 3) & = 1 - Pr(Y \leq 2) = 1 - (Pr(Y=2) + Pr(Y=1)) \\
& = 1 - (0.6^1 \cdot 0.4 + 0.6^0 \cdot 0.4) \\
& = 1 - 0.64 = 0.36 \end{align*}

b) $$\mathbb{E}(Y) = \frac1p = \frac1{0.4} = 2.5$$
On average, we expect the first topping order to be from the second or third customer of each day.


</details>

### Scenario 5
You are working at the icecream shop and notice that you are running low on chocolate flavour. You only have enough to make 2 more orders. If you run out, you will close the store to avoid customers becoming angry over the lack of chocolate. You estimate that the probability of a customer ordering chocolate is 0.27.

a) How many more customers do you expect to serve before you run out and close the store?

b) The store is only open for another 30 minutes anyway, during which period you estimate you can serve one customer per minute. What is the probability that you close the store at closing time without running out of chocolate first?

<details> <summary>▶️ Click to show the solution</summary>

Let $X$ be the number of customers until the second chocolate order. $X$ follows a negative binomial distribution with $k=2$ and $p=0.27$.

a) $$\mathbb{E}(X) = \frac{k}{p} = \frac{2}{0.27} \approx 7.41$$
You expect to serve 7 or 8 customers before running out of chocolate.

b) This time we can model the number of customers ordering chocolate as $Y$, a regular binomial distribution with 30 total customers and probability 0.27 of ordering chocolate. Then we want to find the probability that less than 2 out of 30 customers order chocolate.
\begin{align*}
Pr(Y < 2) & = P(Y=0) + P(Y=1) \\
& = (1-0.27)^30 + 30 \cdot 0.27 \cdot (1-0.27)^{29} \\
& \approx 0.00096
\end{align*}
So there is only a 0.096% that you make it to closing time without needing to close early because you ran out of chocolate.

</details>

### Scenario 6
Let $N$ be the number of toppings ordered in an hour.

a) Given a rate of one customer per minute, and probability of ordering toppings of 0.4, calculate the average rate of toppings per hour. Compute $P(N = 15)$.  

b) We now consider a 15-minute interval. What is the corresponding rate parameter for toppings? What is the probability of no topping orders in this interval?



<details> <summary>▶️ Click to show the solution</summary>

We can model $N$ using a Poisson distribution.

a) The Poisson rate parameter is $\lambda = 60 \times 0.4 = 24$ toppings/hour.

$$Pr(N=15) = \frac{24^{15} e^{-24}}{15!} \approx 0.015$$

b) In a 15-minute interval, the rate becomes $24 \times 0.25 = 6$ toppings/15 minutes (as 15 minutes is a quarter of an hour).

$$Pr(\text{no topping orders in 15 minutes}) = \frac{6^0 e^{-6}}{0!} \approx 0.0025$$

</details>

## Question 2
An airline company accepts 105 bookings for 100-seater planes, knowing that 10\% of passengers fail to turn up for a flight.  Find an expression for the probability that the airline company will be unable to seat all the passengers who arrive for the flight having booked a seat.

<details> <summary>▶️ Click to show the solution</summary>

Let $X$ be the number of passengers that *turn up*. Then $X \sim \mathrm{Bin}(105,0.9)$.

\begin{align*}
			Pr(\text{airline unable to seat all passengers}) &= Pr(X > 100) \\
      &= Pr(X \ge 101)\\
				&= \sum_{x=101}^{105} \binom{105}{x} 0.9^x 0.1^{105-x}\\
				&\approx 0.01672.
		\end{align*}
The probability that the airline is unable to seat all passengers is approximately 0.017.

</details>

## Question 3
Given the probability of successfully winning the jackpot in Oz lotto is 1/45379620, write an expression for the probability of winning the Oz lotto jackpot within a single year (52 weeks) if you purchase a ticket once every week using:


a) The geometric distribution.

b) The binomial distribution.

In both cases, assume that the draw each week is independent of the other draws. Feel free to use R as a calculator to evaluate your expression.

<details> <summary>▶️ Click to show the solution</summary>

a) Let $X \ge 1$ be the number of weeks until we win the Oz lotto. Then $X \sim \mathrm{Geo}(1 / 45379620)$ and "win within a year" = $(X \le 52)$.
\begin{align*}
					Pr(X \le 52) &= \sum_{x=1}^{52} Pr(X = x),\\
						&= \sum_{x=1}^{52}\left(1 - \dfrac{1}{45379620}\right)^{x-1} \times \dfrac{1}{45379620},\\
						&\approx 1.1459 \times 10^{-6}.
				\end{align*}


b) Let $N \ge 0$ be the number of weeks in a year (52 weeks) in which you win the Oz lotto. Then  $N \sim \mathrm{Bin}(52, 1 / 45379620)$ and "win within a year" = $(N \ge 1)$.

\begin{align*}
					Pr(N \ge 1)	&= 1 - Pr(N = 0),\\
						&= 1 - \binom{52}{0} \left(\dfrac{1}{45379620}\right)^0 \left(1 - \dfrac{1}{45379620}\right)^{52},\\
						&= 1 - \left(1 - \dfrac{1}{45379620}\right)^{52},\\
						&\approx 1.1459 \times 10^{-6}.
				\end{align*}

</details>

## Question 5
The number of machine failures per day in a certain plant has a Poisson distribution with a mean of 2. Present facilities at the plant can repair two machines per day. Failures in excess of two are repaired by an outside contractor.


a) What is the probability, on a given day, of having machines repaired by the contractor?

b) What is the probability that the contractor is required at least one day in a 5-day working week? You may assume that the number of failures is independent between days.

<details> <summary>▶️ Click to show the solution</summary>

Let $X$ be the number of machine failures in a day. Then $X \sim \mathrm{Po}(2)$.


a) A contractor is called if three or more machines fail, $X \ge 3$.
				\begin{align*}
					Pr(X \ge 3) &= 1 - Pr(X \le 2),\\
						&= 1 - \sum_{x=0}^2 \dfrac{e^{-2} 2^x}{x!},\\
						&= 1 - e^{-2}\left(\dfrac{2^0}{0!} + \dfrac{2^1}{1!} + \dfrac{2^2}{2!}\right),\\
						&\approx 0.3233.
				\end{align*}

b) Let $Y$ be the number of days in a week that the contractor is required. Then $Y \sim \mathrm{Binom}(5,0.3233)$. Then
				\begin{align*}
					Pr(Y \ge 1) &= 1 - Pr(Y = 0) = 1 - \binom{5}{0} 0.3233^0 (1 - 0.3233)^{5-0} \approx 	0.8581.
				\end{align*}

# Discrete Random Variables in R




#### **R Quick Reference**

`R` supports the following functions for computing distributional quantities and simulating from binomial distributions:

- `dbinom(x, size, prob, log = FALSE)` computes PMF of `Binomial(size,p)`;
- `pbinom(q, size, prob, lower.tail = TRUE, log.p = FALSE)` computes CDF of `Binomial(size,p)`;
- `qbinom(p, size, prob, lower.tail = TRUE, log.p = FALSE)` computes `p`-quantile of `Binomial(size,p)`;
- `rbinom(n, size, prob)` simulates `n` random numbers from `n` independent r.v. following `Binomial(size,p)`.

Here,

- `x`, `q`: The number of successes.
- `size`: The number of Bernoulli trials. If `size = 1`, we obtain the Bernoulli distribution.
- `prob`: The probability of success
- `log`: Whether or not we want to return log PMF.
- `lower.tail`: If `TRUE`, return $\Pr(X \leq x)$ where $X \sim \text{Binomial(size,p)}$. If `FALSE`, return $\Pr(X > x)$.

#### **R Examples**


##### **Example 1**

Toss a fair coin 100 times and count the number of heads. Repeat this entire experiment 10 times using simulation.

<details>
<summary>▶️ Click to show the solution</summary>
We can model the outcome of this experiment by a random variable $Y \sim \text{Binomial}(n,p)$ where $n = 100$ and $p = 0.5$. To simulate random numbers of this r.v., we can use the `rbinom()` function:

```r
rbinom(n = 10, size = 100, p = 0.5)
```
</details>

##### **Example 2**
A fair coin is tossed $10$ times. What is the probability of observing exactly 6 heads?

<details>
<summary>▶️ Click to show the solution</summary>
We can model the outcome of this experiment by a random variable $Y \sim \text{Binomial}(n,p)$ where $n = 10$ and $p = 0.5$. To calculate the probability of observing exactly 6 heads, we can use the PMF of `Binomial(10,0.5)`

$$P(Y = 6) = \binom{10}{6} 0.5^{6}(1-0.5)^{10-6}.$$

We could calculate this by hand, or use `dbinom()` in R.
```r
dbinom(x = 6, size = 10, prob = 0.5)
```
</details>

##### **Example 3**
A fair coin is tossed $10$ times. Compute probability of observing at least 6 heads.


<details>
<summary>▶️ Click to show the solution</summary>
We are interested in computing:

$$\Pr(Y \geq 6).$$

This can be done using the `pbinom()` function in `R`. Note that `pbinom()` always calculates $Pr(Y \leq y)$, so we need to either calculate $1 - Pr(Y \leq 5)$ using the argument `lower.tail = TRUE` or $Pr(Y > 5)$ with the argument `lower.tail = FALSE`.  

```r
1 - pbinom(q = 5, size = 10, prob = 0.5, lower.tail = TRUE)

pbinom(q = 5, size = 10, prob = 0.5, lower.tail = FALSE)
```
</details>

##### **Example 4**

A factory makes 5000 light bulbs an hour (5000 per batch), and each bulb has a 1% chance of being defective. What is the number of defective bulbs that we would not exceed in 95% of the batches?



<details>
<summary>▶️ Click to show the solution</summary>

The number of defective bulbs that we would not exceed in 95% of the batches is $95\%$-quantile of `Binomial(n,p)`, which can be computed using the `qbinom()` function.

## Poisson Distribution






#### **R Quick Reference**

- `dpois(x, lambda, log = FALSE)` computes PMF of `Poisson(lambda)`;

- `x`, `q`: Number of events.

- `lower.tail`: If `TRUE`, return $\Pr(Y \leq x)$; if `FALSE`, return $\Pr(Y > x)$.

## Geometric distribution

#### **R Quick Reference**

`R` supports the following functions for computing distributional quantities and simulating from Poisson distributions:

- `dgeom(x, prob, log)` computes PMF of `Geometric(prob)`;  
- `pgeom(q, prob, lower.tail = TRUE)` computes CDF of `Geometric(prob)`;  
- `qgeom(p, prob)` computes `p`-quantile of `Geometric(prob)`;  
- `rgeom(n, prob)` simulates `n` random numbers from `Geometric(prob)

Importantly, the built-in geometric distribution in R counts the *number of failures* before success. If you need to calculate the total number of trials, you'll need to add 1 to `x`.

### Example: does Brisbane have good coffee?
You have decided to embark on a quest to find the best coffee in Brisbane near uni. While pessimistic about the possibility of finding a truly good coffee shop, you decide that a coffee rated 4.5/5 is acceptably good to end your search. Most coffee shops are pretty average, but a wise friend has indicated that they consider that about 1 in 21 coffee shops meet your standards.

You decide to visit a different shop every morning until you finally find a good one.

1) What is the probability that you visit exactly 6 terrible shops before finding a good one on your 7th day?

2) You realise there are only 15 coffee shops within 10 minutes walk of campus. What is the probability that you find a 4.5/5 coffee spot before needing to extend your search further afield?

3) Use geom() to simulate 1,000 students' quests. Suppose that every "bad" coffee costs you $5.50. What is the average amount of money wasted on bad coffee before finding a good spot?

<details>
<summary>▶️ Click to show the solution</summary>

```r
# Probability of finding a good shop
p_coffee <- 1/21

# 1. Exactly 6 failures (bad shops) before the success
# remember that dgeom calculates number of failures
prob_6_bad <- dgeom(x = 6, prob = p_coffee)
print(prob_6_bad)

# 2. Success within 15 shops total
# This means failures (q) can be anything from 0 to 14
prob_success_14 <- pgeom(q = 14, prob = p_coffee)
print(prob_success_14)

# 3. Financial Simulation
# Simulate number of failures for 1,000 students
failures <- rgeom(n = 1000, prob = p_coffee)

# Calculate wasted money (failures * $5.50)
wasted_money <- failures * 5.50
avg_waste <- mean(wasted_money)

print(avg_waste)
# is this amount worth it?
```

</details>
